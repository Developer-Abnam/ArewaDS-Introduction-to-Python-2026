## Part 1: From class

A list holds many values in order.

````python
scores = [88, 92, 79, 95, 84]
names = ["Ana", "Ben", "Chidi"]
empty = []
print(scores)
print(names)
print(empty)
````

append() adds to the end. The list changes in place.

````python
tasks = []
tasks.append("email")
tasks.append("lab")
tasks.append("gym")
print(tasks)
print(len(tasks))
````

The largest-so-far pattern.

````python
scores = [88, 92, 79, 95, 84]
largest = scores[0]
for score in scores:
    if score > largest:
        largest = score
print(largest)
````

split() turns a string into a list of words.

````python
line = "88 92 79 95 84"
parts = line.split()
print(parts)
scores = []
for part in parts:
    scores.append(int(part))
print(scores)
````

Two names, one list: changing one changes both.

````python
a = [1, 2, 3]
b = a
b.append(4)
print(a)
````

Dictionaries look values up by key. get() avoids a KeyError.

````python
phone = {"Ana": "555-0101", "Ben": "555-0144"}
print("Dee" in phone)
print(phone.get("Dee"))
print(phone.get("Dee", "unknown"))
print(phone.get("Ana", "unknown"))
````

The counting pattern: get(key, 0) + 1.

````python
text = "the cat and the hat and the bat"
counts = {}
for word in text.split():
    counts[word] = counts.get(word, 0) + 1
print(counts)
````

Looping over keys and values together.

````python
stock = {"apples": 7, "kiwis": 7, "plums": 2}
for fruit, count in stock.items():
    print(f"{fruit:8}{count:3}")
````
