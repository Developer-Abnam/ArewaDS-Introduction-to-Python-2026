## Lab 1. `print`, quotes, and escape sequences

### Goals
By the end of this lab you can:
- show text on the screen with `print`
- choose between single quotes, double quotes, and triple quotes
- use the escape sequences `\n`, `\t`, `\"`, and `\'`

### Read this first

**`print(...)`** is a *function*. It shows on the screen whatever you put inside the brackets.

Text in Python is called a **string**. A string must be wrapped in quotes so that Python knows where the text starts and where it ends. You may use `"double quotes"` or `'single quotes'`. They mean exactly the same thing. Just close with the same kind you opened with.

Some characters cannot be typed directly inside a string. For example, how do you type "go to a new line"? For these, Python uses an **escape sequence**: a backslash `\` followed by one character. The backslash says: *"the next character is special"*.

| You type | It means | Example | Output |
|---|---|---|---|
| `\n` | **n**ew line | `"A\nB"` | A, then B on the next line |
| `\t` | **t**ab (jump to the next column) | `"A\tB"` | `A       B` |
| `\"` | a real double quote | `"say \"hi\""` | `say "hi"` |
| `\'` | a real single quote | `'Musa\'s'` | `Musa's` |
| `\\` | a real backslash | `"C:\\data"` | `C:\data` |

````python
# WORKED EXAMPLE 1a: read each comment, THEN run the cell (Shift + Enter).

# 1) print() shows text on the screen. The text goes inside quotes.
print("Ina kwana!")

# 2) You can give print several items, separated by commas.
#    print puts exactly ONE space between them for you.
print("Kano", "Kaduna", "Sokoto")

# 3) \n means "start a new line here". One print, two lines of output.
print("Monday\n\nTuesday")

# 4) \t means "jump to the next tab stop". It is useful for lining up columns.
print("Item\tPrice")
print("Rice\t1500")
print("Beans\t1200")
````

````python
# WORKED EXAMPLE 1b: quotes inside quotes.

# Problem: Python thinks the string ENDS at the second double quote.
#          print("The teacher said "well done"")   <-- this would be a SyntaxError

# Solution A: escape the inner quotes with a backslash.
print("The teacher said \"well done\"")

# Solution B (easier to read): use the OTHER kind of quote on the outside.
print('The teacher said "well done"')
print("It is Bala's book")

# Triple quotes let one string run over several lines. No \n needed.
print("""Dear student,
Welcome to ArewaDS.
Enjoy the lab.""")
````

## Lab 2. Literals and data types

### Goals
By the end of this lab you can:
- say what a **literal** is
- name the three basic types: `int`, `float`, `str`
- use `type(...)` to ask Python what type a value has

### Read this first

A **literal** is a value that you type directly into your code: `42`, `3.14`, `"Kano"`. It *literally* is what it says.

Every value in Python has a **type**. The type decides what you are allowed to do with the value. You can divide two numbers. You cannot divide two names.

| Type | Full name | What it holds | Examples of literals |
|---|---|---|---|
| `int` | integer | whole numbers, no decimal point | `25`, `0`, `-7`, `1000000` |
| `float` | floating-point number | numbers **with** a decimal point | `25.0`, `3.14`, `-0.5` |
| `str` | string | text, always inside quotes | `"Kano"`, `'A'`, `"25"`, `""` |

**The most important idea in this lab:** the quotes decide the type, not the content.

- `25` is a number. You can add to it.
- `"25"` is text that happens to look like a number. You cannot do arithmetic with it.

Think of a phone number like `"08031234567"`. It is made of digits, but you never add two phone numbers together. It is text. It should be a string. (Also, as an `int` it would lose the leading zero!)

````python
# WORKED EXAMPLE 2: asking Python for the type of a value.

# type(...) answers the question "what kind of thing is this?"
print(type(7))          # a whole number           -> int
print(type(7.0))        # has a decimal point      -> float (even though it "equals" 7)
print(type("7"))        # inside quotes            -> str

# The same "25", three different types:
print(25, type(25))
print(25.0, type(25.0))
print("25", type("25"))   # notice: print does not show the quotes, but type() reveals the truth

# Why the type matters: the SAME operator behaves differently.
print(25 + 25)          # numbers: addition
print("25" + "25")      # strings: joined together (more on this in Lab 3)
````

## Lab 3. String operations: `+` and `*`

### Goals
By the end of this lab you can:
- join strings with `+` (this is called **concatenation**)
- repeat a string with `*`
- count characters with `len`
- explain why `"Age: " + 25` fails

### Read this first

You know `+` and `*` from mathematics. With strings they have a different job:

| Expression | Name | Result |
|---|---|---|
| `"Arewa" + "DS"` | concatenation (joining) | `"ArewaDS"` |
| `"ab" * 3` | repetition | `"ababab"` |
| `len("Kano")` | length | `4` |

Two things to remember:

1. **`+` does not add a space for you.** `"Good" + "morning"` gives `"Goodmorning"`. If you want a space, you must put it in yourself: `"Good" + " " + "morning"`.
2. **`+` needs the same type on both sides.** Number + number is fine. String + string is fine. String + number is an error, because Python does not know whether you want to add or to join.

````python
# WORKED EXAMPLE 3: joining and repeating strings.

# + glues strings together, exactly as they are. No space is added.
print("Good" + "morning")           # Goodmorning
print("Good" + " " + "morning")     # Good morning   <- we added the space ourselves

# * repeats a string. string * whole number
print("ha" * 3)                     # hahaha
print("-" * 20)                     # a line of 20 dashes

# A common use: underline a heading with exactly the right number of characters.
heading = "Student Report"
print(heading)
print("=" * len(heading))           # len(heading) is 14, so we get 14 equal signs

# Operators can be combined. * is done before +, just like in mathematics.
print("<" + "=" * 10 + ">")         # <==========>

# len counts EVERY character, including spaces.
print(len("Kano"))                  # 4
print(len("Kano State"))            # 10  (the space counts)
print(len(""))                      # 0   (the empty string)
````

## Lab 4. Variables, naming rules, and constants

### Goals
By the end of this lab you can:
- create a variable and change its value
- explain what the `=` sign means in Python (it is **not** "equals"!)
- tell a legal variable name from an illegal one
- write a constant in the correct style

### Read this first

A **variable** is a **name that points to a value**. Think of a label stuck on a box. The label is the name. What is inside the box is the value.

```python
age = 25
```

Read this line **from right to left**: "take the value `25`, and give it the name `age`".

The `=` sign is called the **assignment operator**. It does not mean "is equal to" as in mathematics. It means **"store the value on the right under the name on the left"**. This is why the following line is perfectly fine in Python, even though it is nonsense in mathematics:

```python
count = count + 1      # take the old value of count, add 1, store the result under the name count again
```

#### Naming rules (Python enforces these, you get a `SyntaxError` if you break them)

| Rule | Legal | Illegal |
|---|---|---|
| Only letters, digits, and underscore `_` | `total_score`, `score2` | `total-score`, `total score`, `score!` |
| Must not start with a digit | `place2` | `2ndPlace` |
| Must not be a Python keyword | `if_value` | `if`, `for`, `class`, `import` |
| Case sensitive | `score` and `Score` are **two different** variables | |

#### Naming conventions (Python does not care, but programmers do)

- Choose names that say what the value **means**: `taxRate`, not `x`.
- Ordinary variables: `camelCase` (as in the slides) or `snake_case`. Pick one and stay with it.
- **Constants**, values that must never change while the program runs, are written in `ALL_UPPERCASE`, for example `TAX_RATE = 0.20`. Python will not stop you from changing it. The capitals are a message to other programmers: *do not touch*.

````python
# WORKED EXAMPLE 4: creating, using, and updating variables.

# 1) Create a variable. Read right to left: "the value 3 gets the name bagsOfRice".
bagsOfRice = 3
pricePerBag = 45000

# 2) Use the variables in an expression. Python replaces each name by its value.
totalCost = bagsOfRice * pricePerBag
print("Total cost:", totalCost)

# 3) A variable can be given a new value at any time. The old value is forgotten.
bagsOfRice = 5
print("Bags now:", bagsOfRice)

# 4) IMPORTANT: totalCost did NOT change by itself. It was computed once, from the old value.
print("Total cost is still:", totalCost)

# 5) To update it, we must compute it again.
totalCost = bagsOfRice * pricePerBag
print("Total cost after recomputing:", totalCost)

# 6) Updating a variable using its own old value. In mathematics this is nonsense; in Python it is normal.
visitors = 10
visitors = visitors + 1      # right side first: 10 + 1 = 11, then store 11 under the name visitors
print("Visitors:", visitors)

# 7) A constant: ALL_UPPERCASE tells the reader "this value must not be changed".
DISCOUNT_RATE = 0.10
print("Discount on the rice:", totalCost * DISCOUNT_RATE)
````

**Stop and think.** Point 4 in the example surprises many beginners. A variable stores a **value**, not a formula. `totalCost` remembered the number `135000`. It did not remember "bags times price". This is different from a spreadsheet.

## Lab 5. Comments and docstrings

### Goals
By the end of this lab you can:
- write an end-of-line comment with `#`
- write a docstring with triple quotes
- explain who comments are written for

### Read this first

Python ignores comments completely. So why write them? Because **programs are read by people**: your teacher, your teammates, and *you*, three months from now, when you have forgotten everything.

| | Comment | Docstring |
|---|---|---|
| Looks like | `# text until the end of the line` | `"""text between triple quotes"""` |
| Where | anywhere | the **first thing** in a program (or function) |
| Purpose | explain one line or one block: *why* is it like this? | describe the whole program: what it does, its input, its output |

**What makes a good comment?** It explains **why**, not **what**.

```python
x = x + 1          # add 1 to x                   <-- useless: the code already says that
```

## Lab 6. Arithmetic expressions and precedence

### Goals
By the end of this lab you can:
- use all seven arithmetic operators
- explain the difference between `/`, `//`, and `%`
- predict the result of an expression using the precedence rules

### Read this first

| Operator | Meaning | Example | Result | Note |
|---|---|---|---|---|
| `+` | addition | `7 + 2` | `9` | |
| `-` | subtraction | `7 - 2` | `5` | |
| `*` | multiplication | `7 * 2` | `14` | |
| `/` | division | `7 / 2` | `3.5` | the result is **always a float**, even `4 / 2` gives `2.0` |
| `//` | quotient (floor division) | `7 // 2` | `3` | how many **whole** times does 2 fit in 7? |
| `%` | remainder (modulus) | `7 % 2` | `1` | what is **left over**? |
| `**` | exponent (power) | `7 ** 2` | `49` | 7 squared |

#### `//` and `%` are partners

You learnt this in primary school: *17 divided by 5 is 3, remainder 2.*

- `17 // 5` gives the `3`
- `17 % 5` gives the `2`
- and they always fit back together: `3 * 5 + 2 == 17`

Real-life picture: you have **17 mangoes** and **5 children**. Each child gets `17 // 5 = 3` mangoes. `17 % 5 = 2` mangoes are left in the basket.

#### Precedence: which operator goes first?

Same as in mathematics (BODMAS / PEMDAS):

| Priority | Operators | Direction |
|---|---|---|
| 1 (first) | `( )` parentheses | |
| 2 | `**` | **right to left** |
| 3 | unary minus, as in `-5` | |
| 4 | `*`  `/`  `//`  `%` | left to right |
| 5 (last) | `+`  `-` | left to right |

**When in doubt, add parentheses.** They cost nothing and make the code easier to read.

````python
# WORKED EXAMPLE 6a: the three kinds of division.

mangoes = 17
children = 5

print(mangoes / children)     # 3.4  true division: always a float
print(mangoes // children)    # 3    each child gets 3 whole mangoes
print(mangoes % children)     # 2    two mangoes are left over

# They always fit back together:
print((mangoes // children) * children + (mangoes % children))   # 17 again

# A very common use of %: is a number even or odd?
print(10 % 2)     # 0 -> no remainder when dividing by 2 -> even
print(7 % 2)      # 1 -> remainder 1 -> odd

# Another common use: converting units.
totalMinutes = 135
print(totalMinutes // 60, "hours and", totalMinutes % 60, "minutes")   # 2 hours and 15 minutes
````

````python
# WORKED EXAMPLE 6b: precedence, step by step.

# Multiplication before addition:
#   10 + 2 * 3
# = 10 + 6
# = 16
print(10 + 2 * 3)

# Parentheses change the order:
#   (10 + 2) * 3
# = 12 * 3
# = 36
print((10 + 2) * 3)

# Same priority -> work from LEFT to RIGHT:
#   20 / 5 * 2
# = 4.0 * 2
# = 8.0
print(20 / 5 * 2)

# A longer one. Do it on paper first!
#   2 + 3 * 4 ** 2 // 5
# = 2 + 3 * 16 // 5        (** first)
# = 2 + 48 // 5            (* and // have equal priority, so left to right: * first)
# = 2 + 9                  (48 // 5 = 9)
# = 11
print(2 + 3 * 4 ** 2 // 5)
````

## Lab 7. Mixed-mode arithmetic and type conversion

### Goals
By the end of this lab you can:
- predict the type of a result when `int` and `float` are mixed
- convert between types with `int()`, `float()`, and `str()`
- explain the difference between **truncating** and **rounding**

### Read this first

**Mixed-mode arithmetic.** When an expression mixes an `int` and a `float`, Python quietly turns the `int` into a `float` first. So the answer is always a `float`. `3 * 2.0` is `6.0`, not `6`.

**Type conversion functions.** Sometimes you must change the type yourself:

| Function | What it does | Example | Result |
|---|---|---|---|
| `int(x)` | makes an integer. From a float it **cuts off** the decimals (truncation). It does **not** round. | `int(3.99)` | `3` |
| | from a string: only works if the string looks like a whole number | `int("33")` | `33` |
| `float(x)` | makes a float | `float(22)`, `float("3.7")` | `22.0`, `3.7` |
| `str(x)` | makes a string, so that you can join it with `+` | `str(99)` | `"99"` |
| `round(x)` | rounds to the nearest integer | `round(3.99)` | `4` |
| `round(x, n)` | rounds to `n` decimal places | `round(3.14159, 2)` | `3.14` |

Think of `int()` as a **knife**: it chops off everything after the decimal point. `int(3.99)` is `3`. If you want the nearest whole number, use `round()`.

````python
# WORKED EXAMPLE 7: converting between types.

# 1) int + float -> float. Python converts the int automatically.
print(5 + 2.0)               # 7.0, not 7
print(type(5 + 2.0))

# 2) int() CUTS, round() ROUNDS.
print(int(9.99))             # 9    the .99 is simply thrown away
print(round(9.99))           # 10

# 3) Strings to numbers. Without conversion, "+" would JOIN the strings.
first = "12"
second = "8"
print(first + second)                  # 128   <- joined as text!
print(int(first) + int(second))        # 20    <- converted, then added

# 4) Numbers to strings. Needed whenever you build a sentence with +.
marks = 87
print("You scored " + str(marks) + " marks")

# 5) A price with kobo: str shows all the decimals the float has.
price = 1500.5
print("Price: N" + str(price))
````

### Read this first

A computer can only store **numbers**. So how does it store the letter `A`? By agreement: everybody agreed that the number **65** stands for `A`, 66 for `B`, and so on. This agreed table is called a **character set**. The classic one is **ASCII**. The modern, worldwide one is **Unicode** (which also contains Hausa hooked letters such as `ɓ`, `ɗ`, `ƙ`, Arabic script, and emoji).

| Function | Direction | Example | Result |
|---|---|---|---|
| `ord(c)` | character → number | `ord('A')` | `65` |
| `chr(n)` | number → character | `chr(65)` | `'A'` |

They are opposites: `chr(ord('A'))` takes you back to `'A'`.

Three facts about the table that are worth remembering:

- The capital letters `A` to `Z` are in order, side by side (65 to 90).
- The small letters `a` to `z` are in order too (97 to 122).
- The digit characters `'0'` to `'9'` are in order too (48 to 57). Note that the **character** `'0'` has code 48. It is not the number 0.

Because the letters are in order, you can do **arithmetic on letters**: "the letter after `a`" is `chr(ord('a') + 1)`.

````python
# WORKED EXAMPLE 10: characters are numbers in disguise.

# 1) From character to code number, and back again.
print(ord('B'))                    # 66
print(chr(66))                     # B
print(chr(ord('B')))               # B   (ord and chr undo each other)

# 2) Letters are stored in alphabetical order, so we can do arithmetic on them.
first = 'a'
code = ord(first)                  # step 1: letter -> number   (97)
code = code + 1                    # step 2: move one place     (98)
print(chr(code))                   # step 3: number -> letter   (b)

# The same three steps, written in one line:
print(chr(ord('a') + 1))           # b
print(chr(ord('a') + 25))          # z, the 26th letter

# 3) The CHARACTER '7' is not the NUMBER 7.
print(ord('7'))                    # 55
print(ord('7') - ord('0'))         # 7   distance from '0' gives the real digit value

# 4) Unicode goes far beyond English.
print(chr(595), chr(1575), chr(8358))   # a Hausa hooked letter, Arabic alif, and the Naira sign
````
